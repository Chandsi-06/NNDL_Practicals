```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=10):
        self.learning_rate = learning_rate
        self.epochs = epochs

    def activation(self, x):
        return 1 if x >= 0 else 0

    def fit(self, X, y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        for _ in range(self.epochs):
            for i in range(len(X)):
                output = np.dot(X[i], self.weights) + self.bias
                prediction = self.activation(output)
                error = y[i] - prediction
                self.weights += self.learning_rate * error * X[i]
                self.bias += self.learning_rate * error

    def predict(self, X):
        return np.array([
            self.activation(np.dot(x, self.weights) + self.bias)
            for x in X
        ])

model = Perceptron(learning_rate=0.1, epochs=10)
model.fit(X, y)

predictions = model.predict(X)

print("Input\tActual\tPredicted")
for i in range(len(X)):
    print(X[i], "\t", y[i], "\t", predictions[i])

print("\nFinal Weights:", model.weights)
print("Final Bias:", model.bias)
```
